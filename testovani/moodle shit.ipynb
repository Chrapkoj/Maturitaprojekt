## Úloha 1 — Test jednoduché funkce (4 b)

Funkce `je_sude(n)` vrací, zda je číslo sudé. Napište testovací třídu `TestJeSude` s testy pro: kladné sudé číslo, kladné liché číslo, nulu a záporné liché číslo.

<details>
<summary>🔎 Očekávaný výstup</summary>


</details>

````python
import unittest


def je_sude(n):
    return n % 2 == 0


class TestJeSude(unittest.TestCase):
    def test_kladne_sude(self):
        self.assertTrue(je_sude(4))

    def test_kladne_liche(self):
        self.assertFalse(je_sude(5))

    def test_nula(self):
        self.assertTrue(je_sude(0))

    def test_zaporne_liche(self):
        self.assertFalse(je_sude(-3))


if __name__ == "__main__":
    unittest.main(argv=["first-arg-is-ignored"], exit=False)
````

## Úloha 2 — Test výjimky (5 b)

Funkce `deleni(a, b)` vrací podíl, při dělení nulou vyvolá `ValueError`. Napište třídu `TestDeleni` a ověřte: běžné dělení, dělení se záporným číslem, dělení nuly (`0 / 5`), dělení nulou a také to, že funkce vrací správný výsledek pro desetinné číslo (`1 / 3`).

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def deleni(a, b):
    if b == 0:
        raise ValueError("Nelze dělit nulou!")
    return a / b


class TestDeleni(unittest.TestCase):
    def test_bezne_deleni(self):
        self.assertEqual(deleni(6, 2), 3)

    def test_deleni_se_zapornym_cislem(self):
        self.assertEqual(deleni(-6, 2), -3)

    def test_deleni_nuly(self):
        self.assertEqual(deleni(0, 5), 0)

    def test_deleni_nulou(self):
        with self.assertRaises(ValueError):
            deleni(5, 0)

    def test_deleni_desetinne_cislo(self):
        self.assertAlmostEqual(deleni(1, 3), 1 / 3)


if __name__ == "__main__":
    unittest.main(argv=["first-arg-is-ignored"], exit=False)
````

## Úloha 3 — Hraniční hodnoty (5 b)

Funkce `omez_do_rozsahu(x, minimum, maximum)` omezí číslo do daného intervalu. Napište třídu `TestOmezDoRozsahu`, která pokryje: hodnotu uvnitř intervalu, hodnotu pod minimem, hodnotu nad maximem a obě hranice intervalu. U testu pod minimem a nad maximem ověřte i jinou hodnotu než v ukázce z výuky.

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def omez_do_rozsahu(x, minimum, maximum):
    return max(minimum, min(x, maximum))


class TestOmezDoRozsahu(unittest.TestCase):
    def test_uvnitr_intervalu(self):
        self.assertEqual(omez_do_rozsahu(5, 1, 10), 5)

    def test_pod_minimem(self):
        self.assertEqual(omez_do_rozsahu(-7, 1, 10), 1)

    def test_nad_maximem(self):
        self.assertEqual(omez_do_rozsahu(42, 1, 10), 10)

    def test_dolni_hranice(self):
        self.assertEqual(omez_do_rozsahu(1, 1, 10), 1)

    def test_horni_hranice(self):
        self.assertEqual(omez_do_rozsahu(10, 1, 10), 10)

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
````

Funkce `je_prestupny_rok(rok)` má vracet, zda je rok přestupný (dělitelný 4, ale ne 100, kromě let dělitelných 400). Napište třídu `TestPrestupnyRok` s testy pro běžný přestupný rok, běžný nepřestupný rok, století a rok dělitelný 400.

Funkce **obsahuje chybu**. Vaše testy ji musí odhalit (alespoň jeden test musí selhat). Funkci neopravujte. Do komentáře pod buňku napište, který test selhal a v čem je chyba.

<details>
<summary>🔎 Očekávaný výstup</summary>

````python
import unittest


def je_prestupny_rok(rok):
    return rok % 4 == 0 and rok % 100 != 0


class TestPrestupnyRok(unittest.TestCase):
    def test_bezny_prestupny_rok(self):
        self.assertTrue(je_prestupny_rok(2024))

    def test_bezny_neprestupny_rok(self):
        self.assertFalse(je_prestupny_rok(2023))

    def test_stoleti_neprestupne(self):
        self.assertFalse(je_prestupny_rok(1900))

    def test_rok_delitelny_400(self):
        self.assertTrue(je_prestupny_rok(2000))

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)

#selze test rokdelitelny 400 protoze rok 2000 neni prestupný
````

## Úloha 5 — Seznamy a výjimky (5 b)

Funkce `soucty_prvku(a, b)` sečte odpovídající prvky dvou seznamů stejné délky, jinak vyvolá `ValueError`. Napište třídu `TestSouctyPrvku` a ověřte: běžné sečtení, prázdné seznamy, seznamy se zápornými čísly, seznamy různé délky a také to, že funkce **nezmění** původní seznamy.

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def soucty_prvku(a, b):
    if len(a) != len(b):
        raise ValueError("Seznamy musí mít stejnou délku!")
    return [x + y for x, y in zip(a, b)]


class TestSouctyPrvku(unittest.TestCase):
    def test_bezne_secteni(self):
        self.assertEqual(soucty_prvku([1, 2, 3], [4, 5, 6]), [5, 7, 9])

    def test_prazdne_seznamy(self):
        self.assertEqual(soucty_prvku([], []), [])

    def test_seznamy_se_zapornymi_cisly(self):
        self.assertEqual(soucty_prvku([-1, 5, -3], [-4, -2, 3]), [-5, 3, 0])

    def test_seznamy_ruzne_delky(self):
        with self.assertRaises(ValueError):
            soucty_prvku([1, 2], [1, 2, 3])

    def test_nemeni_puvodni_seznamy(self):
        a = [1, 2, 3]
        b = [4, 5, 6]
        soucty_prvku(a, b)
        self.assertEqual(a, [1, 2, 3])
        self.assertEqual(b, [4, 5, 6])

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
````

<details>
<summary>🔎 Očekávaný výstup</summary>


</details>

````python
import unittest


class Zasobnik:
    def __init__(self, kapacita=3):
        self.kapacita = kapacita
        self.data = []

    def vloz(self, x):
        if len(self.data) >= self.kapacita:
            raise OverflowError("Zásobník je plný!")
        self.data.append(x)

    def vyjmi(self):
        if not self.data:
            raise IndexError("Zásobník je prázdný!")
        return self.data.pop()

    def je_prazdny(self):
        return len(self.data) == 0

    def velikost(self):
        return len(self.data)


# TODO: Úloha 6
````
